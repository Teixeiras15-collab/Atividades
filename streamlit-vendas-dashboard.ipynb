# **Fase 1: Estrutura Básica e Otimização de Desempenho**

````python
!pip install streamlit
import streamlit as st
import pandas as pd

st.title('Dashboard de Vendas')

@st.cache_data
def carregar_dados():
    # Tentativa de corrigir ParserError: usar sep=';' e on_bad_lines='warn'
    df = pd.read_csv('vendas.csv', sep=';', on_bad_lines='warn')
    return df

# Exemplo de como usar a função para carregar os dados
dados = carregar_dados()
# st.write(dados.head()) # Descomente para visualizar os primeiros dados
````

# **Fase 2: Layout e Filtros Laterais (Interatividade)**


````python
st.sidebar.title('Filtros')

# Obter categorias únicas para o filtro
lista_de_categorias = dados['categoria'].unique().tolist()

# Widget de seleção múltipla para categorias
categorias_selecionadas = st.sidebar.multiselect(
    'Selecione as Categorias',
    options=lista_de_categorias,
    default=lista_de_categorias # Seleciona todas por padrão
)

# Filtrar o DataFrame com base nas categorias selecionadas
dados_filtrados = dados[dados['categoria'].isin(categorias_selecionadas)]

# Exibir os dados filtrados (opcional, para verificar o filtro funcionando)
# st.subheader('Dados Filtrados')
# st.dataframe(dados_filtrados)
````

# **Fase 3: Métricas em Destaque e Visualização de Dados**

````python
import pandas as pd
import streamlit as st

# Assegurar que 'data_hora' seja datetime para manipulação temporal
dados_filtrados['data_hora'] = pd.to_datetime(dados_filtrados['data_hora'])

# Passo 1 e 2: Dividir em colunas e exibir métricas
col1, col2 = st.columns([1, 1])

with col1:
    receita_total = dados_filtrados['valor_venda'].sum()
    st.metric(label='Receita Total', value=f'R$ {receita_total:,.2f}'.replace(',', 'X').replace('.', ',').replace('X', '.'))

with col2:
    total_pedidos = dados_filtrados.shape[0]
    st.metric(label='Total de Pedidos', value=total_pedidos)

# Passo 3: Criar abas
aba1, aba2 = st.tabs(['Evolução Mensal', 'Tabela de Dados'])

with aba1:
    st.subheader('Evolução Mensal da Receita')
    # Agrupar dados por mês e somar o valor de venda
    # Certifique-se de que a coluna de data seja usada corretamente para resampling
    dados_mensais = dados_filtrados.set_index('data_hora').resample('M')['valor_venda'].sum().reset_index()
    dados_mensais['mes_ano'] = dados_mensais['data_hora'].dt.strftime('%Y-%m') # Formato para o eixo X
    st.area_chart(dados_mensais.set_index('mes_ano')['valor_venda'])

with aba2:
    st.subheader('Tabela de Dados Filtrados')
    st.dataframe(dados_filtrados) # Exibe o DataFrame filtrado de forma interativa

    # Preparar dados para download
    csv = dados_filtrados.to_csv(index=False).encode('utf-8')
    st.download_button(
        label='Baixar Dados em CSV',
        data=csv,
        file_name='dados_filtrados.csv',
        mime='text/csv',
    )
````

## Próximos Passos para Deploy na Nuvem (Streamlit Community Cloud)

Com os arquivos `meu_dashboard.py`, `requirements.txt` e `vendas.csv` criados, estamos prontos para fazer o deploy do aplicativo Streamlit. O método mais comum é utilizando o Streamlit Community Cloud em conjunto com um repositório GitHub.

### 1. Criar um Repositório GitHub

Crie um novo repositório público no GitHub. Por exemplo, você pode chamá-lo de `streamlit-vendas-dashboard`.

### 2. Adicionar os Arquivos ao Repositório

Faça upload ou *commit* dos três arquivos (`meu_dashboard.py`, `requirements.txt`, `vendas.csv`) para o seu novo repositório GitHub.

### 3. Deploy no Streamlit Community Cloud

1. Acesse [share.streamlit.io](https://share.streamlit.io/) e faça login com sua conta GitHub.
2. Clique em **'New app'** no canto superior direito.
3. Selecione o repositório que você acabou de criar (`streamlit-vendas-dashboard`).
4. Especifique o arquivo principal como `meu_dashboard.py`.
5. Clique em **'Deploy!'**.

O Streamlit Community Cloud irá automaticamente ler o `requirements.txt` para instalar as dependências e então rodar seu aplicativo. Após alguns minutos, seu dashboard estará online e acessível através de uma URL pública!
